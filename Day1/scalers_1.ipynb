```python
from sklearn.preprocessing import StandardScaler,MinMaxScaler
import numpy as np
import pandas as pd
```

```python
a = np.array([[34, 678, 0.4],
              [45, 786, 0.6],
              [49, 893, 0.9],
              [58, 700, 0.1]])
df = pd.DataFrame(a, columns=['X1','X2','X3'])
df
```

```python
df['X1'].mean(), df['X1'].std()
```

```python
( df['X3'] - df['X3'].mean() )/ df['X3'].std()
```

```python
scaler = StandardScaler().set_output(transform='pandas')
scaler.fit(df) # computes mean and sd of every column
```

```python
df_scl = scaler.transform(df)
df_scl.mean(), df_scl.std()
```

```python
scaler = MinMaxScaler().set_output(transform='pandas')
scaler.fit(df) # computes Min and Max of every column
```

```python
df_scl = scaler.transform(df)
df_scl
```

```python
a = np.array([[34, 678, 0.4],
              [45, 678, 0.6],
              [49, 678, 0.9],
              [58, 678, 0.1]])
df = pd.DataFrame(a, columns=['X1','X2','X3'])
scaler = MinMaxScaler().set_output(transform='pandas')
scaler.fit_transform(df)
```
